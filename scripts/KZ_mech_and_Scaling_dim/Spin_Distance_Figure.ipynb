```python
import matplotlib.pyplot as plt
import numpy as np

import random
```

```python
xvals = np.arange(1,25)

up_list =[]
down_list=[]
for x in xvals:
    r = random.random()
    if r> .1:
        up_list.append(x)
    else:
        down_list.append(x)
```

```python
print(up_list)
```

```python
fig, ax = plt.subplots(figsize=(8,2.5))
ax.scatter(up_list,   np.ones(len(up_list)),   s=100, marker=r"$\uparrow$", color="black")
ax.scatter(down_list, np.ones(len(down_list)), s=100, marker=r"$\downarrow$", color="black")

def bracket(ax, x1, x2, y=1.25, h=0.08, lw=2, color="k"):
    ax.plot([x1, x2], [y, y], color=color, lw=lw)
    ax.plot([x1, x1], [y-h, y+h], color=color, lw=lw)
    ax.plot([x2, x2], [y-h, y+h], color=color, lw=lw)

# highlight ranges
bracket(ax, 1, 5,  y=1.5)
bracket(ax, 11, 15, y=1.5)

# red double-arrow showing distance between bracket centers
c1, c2 = (1+5)/2, (11+15)/2
yline = 1.45
ax.annotate("", xy=(11, yline+.0), xytext=(5, yline),
            arrowprops=dict(arrowstyle="<->", color="red", lw=2))
ax.text((c1+c2)/2, yline+0.05, r"$r$", ha="center", va="bottom", color="red")
ax.text((c1),0, r"$P_n$", ha="center", va="bottom", color="blue")
ax.text((c2), 0, r"$P_n$", ha="center", va="bottom", color="blue")

ax.set_ylim(0.8, 1.6)
ax.set_xlim(-0.5, 16.5)
ax.axis("off")
plt.tight_layout()
plt.show()
```
