```python
import os
import sys
sys.path.append("../../src")
import kibble_zurek as kz
import numpy as np
from tqdm import tqdm
import matplotlib.pyplot as plt 
from matplotlib.pyplot import plot, scatter, loglog, yscale, xscale, xlim, ylim, title
from kibble_zurek import k_vals

from numpy import sqrt, log, exp, array, cos, sin, cosh, tanh, sinh, pi
```

```python
def w(q,g):
    return np.sqrt((g-np.cos(q))**2 + np.sin(q)**2)
def a(q,g):
    return g - cos(q)
def b(q,g):
    return sin(q)
def u(q,g):
   return  (a(q,g)-w(q,g))/sqrt(2*w(q,g)*( w(q,g)-a(q,g)))
def v(q,g):
    return b(q,g)/sqrt(2*w(q,g)*( w(q,g)-a(q,g)))

def state(L,t,g0,g):
    q = k_vals(L)
    u0 = u(q,g0)
    v0 = v(q,g0)
    alpha = a(q,g)
    beta = b(q,g)
    omega = w(q,g)

    u_t = u0*cos(2*omega*t)+1j*(alpha*u0+beta*v0)/omega*sin(2*omega*t)
    v_t= -1j*v0*cos(2*omega*t)+(beta*u0-alpha*v0)/omega*sin(2*omega*t)
    return [u_t,v_t,q]
t= np.linspace(0,1000,100)
s = [state(1000,ti,2,0) for ti in t]

sigmaxsigmax = [kz.D(3,si) for si in s]
plot(sigmaxsigmax)
```
