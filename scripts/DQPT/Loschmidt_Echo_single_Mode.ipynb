```python
def LE(u,v,t):
    return np.abs(u+v*exp(-2*omega*1j*t))**2
def crit_times(omega,ns):
    n = np.arange(0,ns)
    return pi/2/omega * (1+ 2*n)
```

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0,10,1000)

omega = 1
params = [(.55,.45,"green"),
          (1/2,1/2,"blue"),
          (1/4,3/4,"red"),
          (.1,.9,"purple")]

fig, axes = plt.subplots(1,2, figsize=(12,4), sharex=True)

# --- Left: Full Loschmidt Echo ---
for u,v,c in params:
    axes[0].plot(t, LE(u,v,t), color=c, label=rf"$|v^2|={v}$")

axes[0].vlines(crit_times(omega,10), 0, 1,
               color="black", ls="--",
               label=r"$\frac{\pi}{2\omega}(2n+1)$")

axes[0].set_xlim(t[0], t[-1])
axes[0].set_ylim(0,1)
axes[0].set_xlabel("Time")
axes[0].set_ylabel("LE")
axes[0].set_title("Full Loschmidt Echo")
axes[0].legend()

# --- Right: Rate Function ---
for u,v,c in params:
    axes[1].plot(t, -np.log(LE(u,v,t)),
                 color=c, label=rf"$|v^2|={v}$")

axes[1].vlines(crit_times(omega,10), 0, 10,
               color="black", ls="--",
               label=r"$\frac{\pi}{2\omega}(2n+1)$")

axes[1].set_xlim(t[0], t[-1])
axes[1].set_ylim(0,10)
axes[1].set_xlabel("Time")
axes[1].set_ylabel(r"$-\log(\mathrm{LE})$")
axes[1].set_title("Rate Function")
axes[1].legend()

plt.tight_layout()
savefig("LoschmidtEcho.png",dpi=500)
```

```python
import numerical_quenches as nq
import numpy as np
import matplotlib.pyplot as plt
from numpy import log

t_eval = np.linspace(-.5,10,4000)
args = (0,100,1000,0)

fig, axs = plt.subplots(1, 2, figsize=(12,5), sharex=True)

# -------- Left: commensurate --------
Ls1 = [2,10,20,40,100,500]
for L in Ls1:
    s = nq.states(L, t_eval, nq.g_pulse, args)
    IS = nq.groundstate(1000, L)
    axs[0].plot(t_eval,
                -log(abs(nq.overlap(s, IS))**2)/L,
                label=rf"$L={L}$")

axs[0].set_title("Commensurate")
axs[0].set_ylim(0,3)
axs[0].set_ylabel(r"$\lambda(t)$")
axs[0].legend()

# -------- Right: incommensurate --------
Ls2 = [3,5,7,11,101]
for L in Ls2:
    s = nq.states(L, t_eval, nq.g_pulse, args)
    IS = nq.groundstate(1000, L)
    axs[1].plot(t_eval,
                -log(abs(nq.overlap(s, IS))**2)/L,
                label=rf"$L={L}$")

axs[1].set_title("Incommensurate")
axs[1].set_ylim(0,1.5)
axs[1].legend()

for ax in axs:
    ax.set_xlim(t_eval[0], t_eval[-1])
    ax.set_xlabel("Time")

fig.suptitle("Ferromagnetic to Paramagnetic Quench")
plt.tight_layout()
plt.savefig("QuenchSystemSizesSideBySide.png")
plt.show()
```
