Code to generate equilibrium correlations

```python
#Imports
import sys 
sys.path.append("../../../src")
import equilibrium_critical_correlations as eq 
from common_imports import *

folder = "../data/"   
r_max = 25
n = 3

r = np.arange(-r_max,r_max)
if n == 0:
    correlations = array([eq.sigma_general([0,i]) for i in tqdm(r)])
    filename = f"critical_correlation_sigma_rmax{r_max}.npz"
else:
    P = eq.P_n(n)
    correlations = array([eq.P_n_correlations(n,i)-P*P for i in tqdm(r)])
    filename = f"critical_correlation_n{n}_rmax{r_max}.npz"

np.save(folder+filename,correlations)
```
