# This code is for the generation of KZ states in batch so I don't have to keep doing this all the time


```python
import sys
sys.path.append("../../../src")
import kibble_zurek as kz
from common_imports import *
```

```python
from joblib import Parallel, delayed
from tqdm import tqdm
import numpy as np
N = 100
taus = np.logspace(-3, 3.5, N)
L = 500
t = 0
states = Parallel(n_jobs=-1)(
    delayed(kz.state)(L=L, tau=tau, t=tau)
    for tau in tqdm(taus, desc="Computing KZ states")
)
# --- Save everything ---
filename = f"../data/data_KZ_L{L}_t_Tau_N{N}_tau{taus[0]:.1e}-{taus[-1]:.1e}.npz"
np.savez_compressed(filename, taus=taus, L=L, t=t, states=states)

```
