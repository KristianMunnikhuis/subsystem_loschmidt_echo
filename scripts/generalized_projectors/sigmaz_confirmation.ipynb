```python
import sys 
sys.path.append("../../src/")
from project_imports import *
import numerical_quenches as nq 
from numerical_quenches import groundstate
import analytical_quench_states as kz
import TFIM_operators_from_states as op
import kibble_zurek as legacy 
```

```python
def P_array(ns,states):
    return array([[op.Pz_n(ni,s) for s in states] for ni in tqdm(ns)])
```

```python
L = 256
```

```python
taus = np.logspace(-1,3,100)
def g_quench(t,args):
    tau, gf = args
    
    g = t/tau 
    if g > gf:
        return gf 
    else: 
        return g 

gf = 5
states=[]
for tau in tqdm(taus):
    args = (tau,gf)
    t_final = tau*gf
    t_eval = np.linspace(0,t_final,2)
    state = nq.states(L,t_eval,g_quench,args)[1]

    states.append(state)
```

```python
ns = array([1,2,3,4,5,6,7,12])
P = P_array(ns,states)
psi_f =  groundstate(gf,L)
P_gs = P_array(ns,[psi_f])
```

```python
psi_z = groundstate(1000,L)
```

```python
ol = nq.overlap(states,psi_z)
```

```python
x=nq.overlap([psi_z],psi_f)
plot(taus,np.abs(x-ol))
loglog()
```

```python
for Pn, pgs,n in zip(P, P_gs,ns):
    plot(taus,np.abs(Pn-pgs),label=f"n={n}")

plot(taus,4*(2*pi*sqrt(2*taus))**(-1/2)/1.5,"--k")
xlabel(r"$\tau$")
ylabel(r"$\delta^z$")
xlim(taus[0],taus[-1])
ylim(10e-3/2,1)
legend()
title(fr"$\delta^z_n(\tau)$, $g_f= {gf} $")
loglog()
```

```python
for Pn, pgs,n in zip(P, P_gs,ns):
    plot(taus,np.abs(Pn-pgs)/n,label=f"n={n}")

plot(taus,4*(2*pi*sqrt(2*taus))**(-1/2)/1.5/12,"k",label=r"$\sim \frac{1}{\xi}$")
plot(taus,(2*pi*sqrt(2*taus))**(-1),"--k",label=r"$= \frac{1}{\xi^2}$")

xlabel(r"$\tau$")
ylabel(r"$\delta^z$")
xlim(taus[0],taus[-1])
ylim(10e-3/2,1)
legend()
title(fr"$\delta^z_n(\tau)/n$, $g_f= {gf} $")
loglog()
```
