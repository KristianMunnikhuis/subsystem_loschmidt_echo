```python
from tqdm.contrib.concurrent import thread_map  # instead of process_map

def correlator_data(n, rmin, rmax):
    Pn = P_n(n)
    r_points = np.arange(rmin, rmax)
    def _eval(r): return P_n_correlations(n, r) - Pn**2
    data = np.array(thread_map(_eval, r_points, chunksize=32))
    return [data, r_points]

def correlator_scaled(n, rmin, rmax):
    Pn = P_n(n)
    r_points = np.arange(rmin, rmax)
    def _eval(r): return (P_n_correlations(n, r) - Pn**2) / (Pn**2)
    data = np.array(thread_map(_eval, r_points, chunksize=32))
    return [data, r_points]

def correlator_unconnected(n, rmin, rmax):
    r_points = np.arange(rmin, rmax)
    data = np.array(thread_map(lambda r: P_n_correlations(n, r), r_points, chunksize=32))
    return [data, r_points]
```

```python
folder_name = "parallelized_connected_correlations"
os.makedirs(folder_name, exist_ok=True)

rmin, rmax = 50, 150
for n in range(1, 12):
    data, _ = correlator_data(n, rmin, rmax)
    np.save(f"{folder_name}/correlator_data{n}_r{rmin}-{rmax}.npy", data)
    print(f"n = {n} completed!")
```
