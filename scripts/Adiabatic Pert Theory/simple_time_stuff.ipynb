```python
import sys
sys.path.append("../../src")
from common_imports import *
import kibble_zurek as kz
import kblock_ising_model as kb
```

```python
k = kb.k_vals(100)
U = kb.U(k,1)
```

```python
r = np.arange(1,10)
v = np.logspace(-3,2,100)
#v = [100,50,25,10,5,3,1,.5,.1,.05,.01,.0075,.005,.001]
y=[]
for ri in r:   
    xx= kb.sigma_general([0,ri],U,k)
    
    x=[]
    for vi in v:
        state = kz.state(100,1/vi,0)

        x+= [xx-kz.sigma_general([0,ri],state)]

    x= array(x)
    y.append(x)
    plot(v,x,label=f"{ri}")
legend()
```
