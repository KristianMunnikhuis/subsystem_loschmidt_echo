# This notebook will generate data for comparing correlations of $x$ with $z$

```python
import sys
sys.path.append("../../src")
from equilibrium_z_projectors import *

from common_imports import *
```

```python
def  generate_data(n):
    data = []
    P0 = P_n(n)
    for r in tqdm(range(1,501,5)):
        data.append(P_n_correlations(n,r)-P0*P0)
    data = array(data)
    return data
```

```python
x = generate_data(7)
```

```python
for n in range(1,10):
    data = generate_data(n)

    filename = f"PZPZ_connected_correlation_functions/PZPZ_{n}.npy"
    np.save(filename,data)

    print(f"Complete! \n, now starting {n+1}")
```
