```python
foldername = "PZPZ_connected_correlation_functions/"

def data(n):
    return np.load(foldername+f"PZPZ_{n}.npy")

r = np.arange(1,201)
```

```python
n = np.arange(1,9)
data_list = [data(ni) for ni in n]
```

```python
plt.figure(figsize=(5,5))  # width = 6 inches, height = 6 inches

for corr,ni in zip(data_list,[1,2,3,5,7,8]):
    scatter(r,corr*r*r,label=f"n = {ni}",s=5)

#plot(r,r**(1/8)/50,"--k",label=r"$r^{-2}$")
loglog()
legend()
xlim(50,r[-1])
title(r"$r^{2}\times\langle \psi_0 |P^z_n(0) P^z_n(r)| \psi_0  \rangle_{cc}$")
plt.xlabel(r"Distance $r$")
plt.ylabel(r"$r^{2}\langle P_n P_n \rangle _{cc}$")

```
