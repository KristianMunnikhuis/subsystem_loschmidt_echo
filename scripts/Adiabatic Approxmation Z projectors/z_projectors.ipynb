```python
import sys
sys.path.append("../../src")
import kibble_zurek as kz
import time_evolved_z_correlators as zc
from common_imports import *
import equilibrium_z_projectors as ez
```

```python
L = 50
tau = 10
ts = np.linspace(-tau,tau,20)
states = [kz.state(256,tau,t) for t in tqdm(ts)]
```

```python
def helperz(states,n):
    return array([zc.P_n(n,s) for s in states])
def helperx(states,n):
    return array([kz.P_n(n,s) for s in states])
dats = [helperz(states,n)-ez.P_n(n) for n in np.arange(1,10)]
for dat in dats:
    plot(ts/tau,dat)
    xlim(ts[0]/tau,ts[-1]/tau)
```

```python
taus = np.logspace(-2,3,100)
states = [kz.state(512,tau,tau) for tau in tqdm(taus)]
```

```python
dats = [helperz(states,n)-2.**(-n) for n in np.arange(1,10)]
```

```python
for n,dat in enumerate(dats):
    plot(taus,dat,label=f"n={n+1}")
    xlim(taus[0],taus[-1])
loglog()
legend()
ylabel(r"$\delta_n(\tau)$")
title("Z projectors, quench to g = 0")
xlabel(r"$\tau_Q$")
```
