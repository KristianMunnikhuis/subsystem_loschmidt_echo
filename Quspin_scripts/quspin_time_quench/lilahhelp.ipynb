```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
H0=np.array([[0,1],[1,0]])
E, V = np.linalg.eigh(H0)
print(f"H0={H0}")
print("\n")
print(f"E={E}")
print("\n")
V1 = V[:,0]
V2 = V[:,1]
print(f"V1= {V1}")
print(f"V2= {V2}")


print(f"U= {V}")

print("\n Now let's see how this makes sense. Let's act V1,V2 on H")

print(r"$H|V1=$"+f" {H0@V1}")
print(r"$H|V2>=$"+f" {H0@V2}")

print("\n More Generally:")

print(f"<V1|H|V1>= {np.vdot(V[:,0],H0@V[:,0])} ")
print(f"<V2|H|V2>= {np.vdot(V[:,1],H0@V[:,1])} ")

#plt.imshow(abs(H0))
#plt.colorbar()
```

```python
print("We can use U to diagonalize the hamiltonian")
print(f"\nUT H0 U = \n {np.conj(V).T@H0@V}")

print("\n We also can transform our eigenvectors")

print(f"= {V.T@V1}")
print(f"")
print(f" = {V.T@np.array([1j,-1j])}")

print("\n Finally, we can also verify explicitly the relation that <O> is basis indepdent.")
```
